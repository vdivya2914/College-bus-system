```python
# College Bus Tracking System

buses = {
    "B01": {
        "driver": "Ravi",
        "route": "Kurnool - College",
        "location": "Kurnool",
        "status": "On Route"
    },
    "B02": {
        "driver": "Suresh",
        "route": "Nandyal - College",
        "location": "Nandyal",
        "status": "On Route"
    },
    "B03": {
        "driver": "Ramesh",
        "route": "Banaganapalle - College",
        "location": "Banaganapalle",
        "status": "Stopped"
    }
}


def show_buses():
    print("\n===== ALL COLLEGE BUSES =====")

    for bus_id, bus in buses.items():
        print("\nBus ID:", bus_id)
        print("Driver:", bus["driver"])
        print("Route:", bus["route"])
        print("Location:", bus["location"])
        print("Status:", bus["status"])


def track_bus():
    bus_id = input("\nEnter Bus ID: ").upper()

    if bus_id in buses:
        bus = buses[bus_id]

        print("\n===== BUS DETAILS =====")
        print("Bus ID:", bus_id)
        print("Driver:", bus["driver"])
        print("Route:", bus["route"])
        print("Current Location:", bus["location"])
        print("Status:", bus["status"])
    else:
        print("Bus not found!")


def update_location():
    bus_id = input("\nEnter Bus ID: ").upper()

    if bus_id in buses:
        location = input("Enter new location: ")
        status = input("Enter status (On Route/Stopped): ")

        buses[bus_id]["location"] = location
        buses[bus_id]["status"] = status

        print("Bus location updated successfully!")
    else:
        print("Bus not found!")


while True:

    print("\n==============================")
    print(" COLLEGE BUS TRACKING SYSTEM")
    print("==============================")

    print("1. Show All Buses")
    print("2. Track Bus")
    print("3. Update Bus Location")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        show_buses()

    elif choice == "2":
        track_bus()

    elif choice == "3":
        update_location()

    elif choice == "4":
        print("Thank you!")
        break

    else:
        print("Invalid choice!")
```
